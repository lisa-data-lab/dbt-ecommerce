```python
# Source catalog and schema
source_catalog = "landing"
source_schema = "ecommerce"

# Base path in the volume to store Parquet files
volume_base_path = "/Volumes/bronze_dev/ecommerce/volumes/raw_data/"

# Get all tables in the source schema
tables = [t.name for t in spark.catalog.listTables(f"{source_catalog}.{source_schema}")]

# Loop through each table and write to Parquet
for table_name in tables:
    df = spark.table(f"{source_catalog}.{source_schema}.{table_name}")
    
    # Target path in volume
    volume_path = f"{volume_base_path}{table_name}/"
    
    # Write as Parquet
    df.write.mode("overwrite").parquet(volume_path)
    
    print(f"Table {table_name} written as Parquet to {volume_path}")
```
