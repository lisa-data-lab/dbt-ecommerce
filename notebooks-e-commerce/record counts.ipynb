```python
# List all tables in your schema
tables = spark.sql("SHOW TABLES IN landing.ecommerce").collect()

counts = []
total = 0

for t in tables:
    name = t.tableName
    count = spark.table(f"landing.ecommerce.{name}").count()
    counts.append((name, count))
    total += count

# Create a summary DataFrame
result = spark.createDataFrame(counts, ["table_name", "record_count"])

from pyspark.sql import Row
total_row = spark.createDataFrame([Row(table_name="TOTAL", record_count=total)])

final_result = result.union(total_row)

display(final_result)
```
