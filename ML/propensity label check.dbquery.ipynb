```sql
# =========================
# 1️⃣ Imports
# =========================
import pandas as pd
from pyspark.sql import SparkSession

# =========================
# 2️⃣ Load features and CLV proxy
# =========================
spark = SparkSession.builder.getOrCreate()

# Load propensity features
df_features = spark.table("analytics_ml.int_features.features_propensity").toPandas()

# Load CLV proxy and join
df_clv = spark.table("analytics_ml.int_features.features_clv_proxy")[["customerid","clv_proxy"]].toPandas()
df = df_features.merge(df_clv, on="customerid", how="left")

# =========================
# 3️⃣ Prepare features (X) and label (y)
# =========================
feature_cols = [
    "total_orders_base",
    "avg_order_value",
    "days_since_last_purchase",
    "count_email_open",
    "count_email_click",
    "count_sms_click",
    "count_push_open",
    "clv_proxy"
]

X = df[feature_cols]
y = df["label"]

# =========================
# 4️⃣ Check label distribution
# =========================
print("Label distribution:")
print(y.value_counts())

# Optional: check ratio
print("\nLabel ratio (percentage):")
print(y.value_counts(normalize=True) * 100)
```
