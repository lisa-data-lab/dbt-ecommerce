```sql
SELECT propensity_segment,
       SUM(expected_revenue) AS total_expected_revenue,
       COUNT(*) AS num_customers,
       -- New column for custom sorting
       CASE propensity_segment
           WHEN 'High' THEN 1
           WHEN 'Medium' THEN 2
           WHEN 'Low' THEN 3
           ELSE 99
       END AS segment_order
FROM analytics_ml.int_features.propensity_scores_segmented
GROUP BY propensity_segment
ORDER BY segment_order; -- ORDER BY the custom column instead of revenue
```
