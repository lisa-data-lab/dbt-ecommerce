```sql
CREATE OR REPLACE TABLE analytics_ml.int_features.features_propensity_3 AS
SELECT
    c.customerid,
    c.orders as total_orders_base,
    c.avg_order_value,
    c.days_since_last_purchase,

    COUNT(DISTINCT CASE WHEN j.event_type = 'email_open' THEN j.event_id END) AS count_email_open,
    COUNT(DISTINCT CASE WHEN j.event_type = 'email_click' THEN j.event_id END) AS count_email_click,
    COUNT(DISTINCT CASE WHEN j.event_type = 'sms_click' THEN j.event_id END) AS count_sms_click,
    COUNT(DISTINCT CASE WHEN j.event_type = 'push_open' THEN j.event_id END) AS count_push_open,

    CASE WHEN MAX(i.order_date) >= date_sub('2011-02-02', 30) THEN 1 ELSE 0 END AS label
FROM gold_dev.dbt_lzeng_ecommerce_int.int_customer_base_metrics c
LEFT JOIN gold_dev.dbt_lzeng_ecommerce_staging.stg_journey_events j ON c.customerid = j.customer_id
LEFT JOIN gold_dev.dbt_lzeng_ecommerce_int.int_customers_orders i ON c.customerid = i.customer_id
GROUP BY c.customerid, c.orders, c.avg_order_value, c.days_since_last_purchase;
```
