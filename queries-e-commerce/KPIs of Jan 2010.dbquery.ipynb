```sql
SELECT
  '2010-01-01' AS period_start_date,
  '2010-01-31' AS period_end_date,
  
  ROUND(AVG(DAU), 0) AS Avg_Daily_Active_Users,
  ROUND(AVG(WAU), 0) AS Avg_Weekly_Active_Users,
  ROUND(AVG(MAU), 0) AS Avg_Monthly_Active_Users,

  SUM(Total_Revenue) AS PTD_Total_Revenue,
  SUM(Total_Orders) AS PTD_Total_Orders,

  ROUND(AVG(CVR), 2) AS PTD_Avg_Conversion_Rate_pct, -- Uses the pre-calculated CVR column
  ROUND(AVG(AOV), 2) AS PTD_Avg_Order_Value,       -- Uses the pre-calculated AOV column
  ROUND(AVG(ROAS), 2) AS PTD_Avg_ROAS              -- Uses the pre-calculated ROAS column

FROM gold_dev.dbt_lzeng_ecommerce_mart.mart_daily_kpis 
WHERE date BETWEEN '2010-01-01' AND '2010-01-31';
```
