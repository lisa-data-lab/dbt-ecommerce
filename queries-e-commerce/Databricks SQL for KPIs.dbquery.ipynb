```sql
SELECT
  -- 1. Time Period Definition (Last Full Month)
  DATE_TRUNC('month', DATE_SUB(CURRENT_DATE(), 1)) AS period_start_date,
  DATE_TRUNC('month', CURRENT_DATE()) AS period_end_date_exclusive,
  
  -- 2. User Activity (Averaged over the period)
  ROUND(AVG(DAU), 0) AS Avg_Daily_Active_Users,
  ROUND(AVG(WAU), 0) AS Avg_Weekly_Active_Users,
  ROUND(AVG(MAU), 0) AS Avg_Monthly_Active_Users,

  -- 3. Revenue & Orders (Aggregated Totals)
  SUM(Total_Revenue) AS PTD_Total_Revenue,
  SUM(Total_Orders) AS PTD_Total_Orders,

  -- 4. Conversion Rate (CVR: Total Conversions / Total Impressions)
  ROUND(
    (SUM(Total_Conversions) * 100.0) / NULLIF(SUM(Total_Impressions), 0),
    2
  ) AS PTD_Conversion_Rate_pct,

  -- 5. Average Order Value (AOV: Total Revenue / Total Orders)
  ROUND(
    SUM(Total_Revenue) / NULLIF(SUM(Total_Orders), 0),
    2
  ) AS PTD_Avg_Order_Value,

  -- 6. Return on Ad Spend (ROAS: Total Revenue / Total Spend)
  ROUND(
    SUM(Total_Revenue) / NULLIF(SUM(Total_Spend), 0),
    2
  ) AS PTD_ROAS

FROM dbt_lzeng_ecommerce_mart.mart_daily_kpis
-- Filter for the Last Full Month
WHERE date >= DATE_TRUNC('month', DATE_SUB(CURRENT_DATE(), 1))
  AND date < DATE_TRUNC('month', CURRENT_DATE());
```
