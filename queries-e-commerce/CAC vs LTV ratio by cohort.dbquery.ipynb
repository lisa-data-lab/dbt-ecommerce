```sql
SELECT
    cohort_month,
    COUNT(customerid) AS total_customers,
    SUM(acquisition_cost) AS total_cac,
    SUM(lifetime_value) AS total_ltv,
    CASE 
        WHEN SUM(lifetime_value) > 0 THEN SUM(acquisition_cost) / SUM(lifetime_value)
        ELSE NULL
    END AS cac_ltv_ratio
FROM gold_dev.dbt_lzeng_ecommerce_mart.mart_customer_360
WHERE cohort_month IS NOT NULL
GROUP BY cohort_month
ORDER BY cohort_month;
```
