```sql
CREATE OR REPLACE TABLE analytics_ml.int_features.features_clv_proxy AS
SELECT
    customerid,
    total_orders_base,
    avg_order_value,
    days_since_last_purchase,
    -- Basic revenue metric
    total_orders_base * avg_order_value AS gross_revenue_base,

    -- Simple recency weighting: newer purchasers likely more valuable
    CASE 
        WHEN days_since_last_purchase <= 30 THEN 1.2
        WHEN days_since_last_purchase <= 90 THEN 1.0
        WHEN days_since_last_purchase <= 180 THEN 0.8
        ELSE 0.6
    END AS recency_weight,

    -- Weighted CLV proxy
    (total_orders_base * avg_order_value) *
    CASE 
        WHEN days_since_last_purchase <= 30 THEN 1.2
        WHEN days_since_last_purchase <= 90 THEN 1.0
        WHEN days_since_last_purchase <= 180 THEN 0.8
        ELSE 0.6
    END AS clv_proxy

FROM analytics_ml.int_features.features_propensity;
```
